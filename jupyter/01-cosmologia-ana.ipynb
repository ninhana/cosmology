## 1.1 — Conceitos básicos

A cosmologia busca estudar o Universo como um todo, tentando assim compreender sua origem física e sua evolução a partir das leis da Física. A força gravitacional é a interação mais forte entre esses objetos macroscópicos. Estudamos essa dinâmica do universo usando a relatividade geral de Einsten. 

O princípio cosmológico estabelece que em escalas
suficientemente grandes, o Universo é:

- **Homogêneo:** possui as mesmas propriedades
  estatísticas em diferentes posições.
- **Isotrópico:** possui as mesmas propriedades
  estatísticas em diferentes direções.

A partir dessas hipóteses, temos que a evolução cosmológica pode ser definida pelo fator de escala $a(t)$ e pelo parâmetro de curvatura espacial $k$.

O parâmetro $k$ pode assumir três valores:

- $k=0$: espaço plano;
- $k=+1$: curvatura espacial positiva;
- $k=-1$: curvatura espacial negativa.


### Métrica FLRW

Estudamos esses modelos usando a métrica de Friedmann-Lemaître-Robertson-Walker
(FLRW), nas coordenadas $(t,r,\theta,\phi)$, que é dada por

$$
ds^2=-dt^2+a^2(t)\left[
\frac{dr^2}{1-kr^2}
+r^2d\theta^2
+r^2\sin^2\theta\,d\phi^2
\right].
$$

Como convenção usamos $(-,+,+,+)$ e $c=1$.

O elemento de linha
também pode ser escrito como

$$
ds^2=g_{\mu\nu}dx^\mu dx^\nu.
$$

Comparando as duas expressões, identificamos
os componentes do tensor métrico $g_{\mu\nu}$. Vamos construí-la usando cálculo simbólico.

```python

# coordenadas do espaço-tempo
t, r, theta, phi = sp.symbols('t r theta phi', real=True)

# parâmetro de curvatura
k = sp.symbols('k', real=True)

# fator de escala
a = sp.Function('a')(t)

#tensor métrico
g = sp.diag(-1,a**2/(1-k*r**2),a**2*r**2,a**2*r**2*sp.sin(theta)**2)

display(g)
```


Para conseguirmos calcular os símbolos de Christoffel, precisamos
determinar a métrica inversa $g^{\mu\nu}$, que é definida por:

$$
g_{\mu\nu}g^{\nu\lambda}
=\delta_\mu^{\ \lambda},
$$

onde $\delta_\mu^{\ \lambda}$ representa o
delta de Kronecker.

Como a métrica FLRW é diagonal, sua inversa
é obtida invertendo os elementos da diagonal.


```python
# cálculo da métrica inversa
gI = sp.simplify(g.inv())

display(gI)
```


### Símbolos de Christoffel

Os símbolos de Christoffel associados à
conexão de Levi-Civita são definidos por

$$
\Gamma^\alpha_{\mu\nu}
=
\frac{1}{2}g^{\alpha\beta}
\left(
\partial_\mu g_{\beta\nu}
+\partial_\nu g_{\beta\mu}
-\partial_\beta g_{\mu\nu}
\right).
$$

Os índices assumem os valores $0,1,2,3$,
correspondentes às coordenadas

$$
x^\mu=(t,r,\theta,\phi).
$$

Para calcular os símbolos de Christoffel,
utilizamos o tensor métrico $g_{\mu\nu}$,
sua inversa $g^{\mu\nu}$ e as derivadas
parciais dos componentes da métrica.


```python

# definindo as coord
X = [t, r, theta, phi]

# função que calcula os símbolos de christoffel
def Gamma(alpha, mu, nu):
    resultado = 0

    for beta in range(4):
        resultado += (gI[alpha, beta] / 2 * sp.diff(g[beta, nu], X[mu])
                + sp.diff(g[beta, mu], X[nu]) - sp.diff(g[mu, nu], X[beta]))

    return sp.simplify(resultado)
```

```python

adot = sp.Symbol(r'\dot{a}')
addot = sp.Symbol(r'\ddot{a}')

display(adot)
display(addot)
```

```python

from IPython.display import display, Math

contador = 0

for alpha in range(4):
    for mu in range(4):
        for nu in range(4):

            resultado = sp.simplify(Gamma(alpha, mu, nu))

            if resultado != 0:

                contador += 1

                resultado = resultado.subs(sp.diff(a, t), adot
                )
                
                display(Math(rf"\Gamma^{{{alpha}}}_{{{mu}{nu}}}"rf" = {sp.latex(resultado)}"
                ))

print("Total de símbolos não nulos:", contador)
```

## Tensor de Ricci

Através dos símbolos de Christoffel podemos obter o tensor de Ricci, que é calculado através da contração do tensor de Riemann e que está relacionado à curvatura do espaço-tempo.

$$
R_{ \mu \nu}
=
\partial_\lambda\Gamma^\lambda_{\mu\nu}
-\partial_\nu\Gamma^\lambda_{\mu\lambda}
+\Gamma^\lambda_{\mu\nu}\Gamma^\sigma_{\lambda\sigma}
-\Gamma^\sigma_{\mu\lambda}\Gamma^\lambda_{\nu\sigma}.
$$

Os índices repetidos $\lambda$ e $\sigma$
representam somatórios de $0$ até $3$.

Os únicos componentes não nulos do tensor de Ricci são $R_{00}$, $R_{11}$, $R_{22}$ e $R_{33}$.

```python

from IPython.display import display, Math

X = [t, r, theta, phi]

# símbolos de christoffel
G = [
    [
        [Gamma(alpha, mu, nu) for nu in range(4)]
        for mu in range(4)
    ]
    for alpha in range(4)
]

Ricci = sp.zeros(4, 4)

# calculando todos os componentes
for mu in range(4):
    for nu in range(4):

        T1 = sum(sp.diff(G[l][mu][nu], X[l])
            for l in range(4)
        )

        T2 = -sum(
            sp.diff(G[l][mu][l], X[nu])
            for l in range(4)
        )

        T3 = sum(
            G[l][mu][nu] * G[s][l][s]
            for l in range(4)
            for s in range(4)
        )

        T4 = -sum(
            G[s][mu][l] * G[l][nu][s]
            for l in range(4)
            for s in range(4)
        )

        Ricci[mu, nu] = sp.simplify(
            T1 + T2 + T3 + T4
        )

# somente os componentes não nulos
for mu in range(4):
    for nu in range(4):

        resultado = Ricci[mu, nu]

        if resultado != 0:

            resultado_ponto = resultado.xreplace({sp.diff(a, t, 2): addot,sp.diff(a, t): adot})

            display(Math(
                rf"R_{{{mu}{nu}}} = "
                + sp.latex(resultado_ponto)
            ))
```

## Escalar de Ricci

O escalar de Ricci é a quantidade escalar que define a curvatura do espaço-tempo, que é obtido através da contração do tensor de Ricci
com a métrica inversa:

$$
R=g^{\mu\nu}R_{\mu\nu}.
$$

Como já sabemos que a métrica de FLRW e o tensor de Ricci tem as componentes não nulas dadas pela diagonal principal, a contração pode ser escrita por:

$$
R=g^{00}R_{00} +g^{11}R_{11}+g^{22}R_{22}+g^{33}R_{33}.
$$

```python

from IPython.display import display, Math

termos_R = []

# calculando separadamente os termos da contração

for i in range(4):

    termo = sp.simplify(gI[i,i] * Ricci[i,i])
    termos_R.append(termo)

    display(Math(
        rf"g^{{{i}{i}}}R_{{{i}{i}}} = "
        + sp.latex(com_pontos(termo))
    ))
```

Dessa forma, o escalar de Ricci é dado por:

```python

#soma dos componentes da contração
R_escalar = sp.simplify(sum(termos_R))

# Expandindo para visualizar cada contribuição
R_expandido = sp.expand(R_escalar)

# Exibindo R com a notação de ponto
display(
    sp.Eq(
        sp.Symbol('R'),
        com_pontos(R_expandido)
    )
)
```

## Tensor de Einsten

O tensor de Einstein relaciona o tensor de Ricci,
o escalar de Ricci e o tensor métrico, e é dado por:

$$
G_{\mu\nu}
=
R_{\mu\nu}-\frac12 Rg_{\mu\nu}.
$$

Aqui o tensor de Einsten támbem será diagonal.


```python

G = Ricci - sp.Rational(1, 2)*R_escalar*g

#simplificação dos componentes
G = G.applyfunc(sp.simplify)

# Exibição dos componentes diagonais
for i in range(4):

    resultado = com_pontos(G[i, i])

    display(
        sp.Eq(
            sp.Symbol(f'G_{{{i}{i}}}'),
            resultado
        )
    )
```

## Tensor energia-momento de um fluido perfeito

Para conseguirmos representar a parte da matéria e energia do Universo, consideramos um fluido perfeito, onde seu tensor energia-momento é dado por:

$$
T_{\mu\nu} = (\rho+p)u_\mu u_\nu+pg_{\mu\nu},
$$

onde:

- $\rho(t)$ representa a densidade de energia;
- $p(t)$ representa a pressão;
- $u^\mu$ representa a quadrivelocidade;
- $g_{\mu\nu}$ é o tensor métrico FLRW.

Considerando um referncial onde o fluido está em repouso e considerando $c=1$ temos:

$$
u^\mu=(1,0,0,0).
$$

O quadrivetor covariante é dado por:

$$
u_\mu=g_{\mu\nu}u^\nu.
$$

```python

u_contra = sp.Matrix([1, 0, 0, 0])

u_cov = sp.simplify(g * u_contra)

display(u_contra)
display(u_cov)
```

Agora podemos contruir o tensor $T_{\mu\nu}$:

```python

rho = sp.Function('rho')(t)
p = sp.Function('p')(t)

# tensor energia-momento
Tfluido = ((rho + p) * (u_cov * u_cov.T) + p * g)

Tfluido = Tfluido.applyfunc(sp.simplify)

# Exibindo os componentes diagonais
for i in range(4):
    display(
        sp.Eq(
            sp.Symbol(f'T_{{{i}{i}}}'), Tfluido[i, i]
        )
    )
```

## Equações de Friedmann

Obtemos as equações de Friedmann são obtidas através das equações de campo de Eisnten, considerando a métrica FLRW e um fluido perfeito:

$$
G_{\mu\nu} +\Lambda g_{\mu\nu} = \kappa T_{\mu\nu}
$$

onde $\Lambda$ é a constante cosmológica e
$\kappa=8\pi G_N/c^4$.


### Primeira equação de Friedmann
Considerando a componente temporal $\mu=\nu=0$ obtemos a primeira equação de Friedmann.

```python

# Constantes da equação de Einstein
Lambda, kappa = sp.symbols('Lambda kappa', real=True)

# Componente 00 da equação de Einstein
lado_esquerdo = sp.simplify(
    G[0, 0] + Lambda*g[0, 0]
)

lado_direito = kappa*Tfluido[0, 0]

# Exibindo a primeira equação de Friedmann
display(
    sp.Eq(
        com_pontos(lado_esquerdo),
        lado_direito
    )
)
```


### Resultado da primeira equação de Friedmann

A partir da componente temporal da equação
de campo de Einstein,obtemos

$$
\boxed{
3\frac{\dot a^2}{a^2}
+\frac{3k}{a^2}
-\Lambda=\kappa\rho.
\tag{2a}
}
$$

Essa equação relaciona a evolução do fator de escala $a(t)$ com a densidade de energia $\rho$, a curvatura espacial $k$ e a constante cosmológica $\kappa$.

### Segunda equação de Friedmann

Para obter a segunda equação de Friedmann,
consideramos a componente espacial como $\mu=\nu=1$.


```python


# lado esquerdo
lado_esquerdo_11 = sp.simplify(
    G[1,1] + Lambda*g[1,1]
)

# lado direito
lado_direito_11 = sp.simplify(
    kappa*Tfluido[1,1]
)

display(
    sp.Eq(
        com_pontos(lado_esquerdo_11),
        com_pontos(lado_direito_11)
    )
)
```

```python

# dividindo ambos os lados por g_11
esquerda_simplificada = sp.simplify(
    lado_esquerdo_11 / g[1,1]
)

direita_simplificada = sp.simplify(
    lado_direito_11 / g[1,1]
)

# exibindo
display(
    sp.Eq(
        com_pontos(esquerda_simplificada),
        com_pontos(direita_simplificada)
    )
)
```


### Segunda equação de Friedmann

A partir da componente radial da equação
de campo de Einstein, obtemos

$$
\boxed{
-2\frac{\ddot a}{a}
-\frac{\dot a^2}{a^2}
-\frac{k}{a^2}
+\Lambda=\kappa p
}
\tag{2b}
$$

Essa equação relaciona a evolução do fator
de escala $a(t)$ e sua aceleração com a
pressão $p$ do fluido, a curvatura espacial
$k$ e a constante cosmológica $\Lambda$.


## Equação de energia momento

Somando as equaçãoes $(2a)$ e $(2b)$ e isolando a segunda derivada de $a(t)$:

```python

from IPython.display import display, Math

# primeira equação
F1 = sp.simplify(G[0,0] + Lambda*g[0,0])

# segunda equação
F2 = sp.simplify((G[1,1] + Lambda*g[1,1]) / g[1,1])

soma = sp.simplify(F1 + F2)

display(Math(
    r'\kappa(\rho+p) = '+ sp.latex(com_pontos(soma))
))

# isolando a segunda derivada de a(t)
solucao_addot = sp.solve(sp.Eq(soma, kappa*(rho+p)),sp.diff(a,t,2)
)[0]

display(Math(
    r'\ddot{a} = '+ sp.latex(com_pontos(solucao_addot))
))
```

Derivando a equação $(2a)$:

```python

derivada_F1 = sp.factor(sp.diff(F1, t))

display(Math(
    r'\kappa\dot{\rho} = '+ sp.latex(com_pontos(derivada_F1))
))
```

Derivando a segunda equação $(2b)$

```python

derivada_F1 = sp.factor(sp.diff(F1, t))

display(Math(r'\kappa\dot{\rho} = '+ sp.latex(com_pontos(derivada_F1))))
```

Substituindo a expressão encontrada da segunda derivada de $a(t)$:

```python

# substituindo a segunda derivada de a(t)
rhodot_obtido = sp.simplify(derivada_F1.subs(sp.diff(a,t,2),solucao_addot) / kappa)

# exibindo o resultado 
display(Math(r'\dot{\rho} = '+ sp.latex(com_pontos(rhodot_obtido))))
```


### Resultado da equação de conservação

Derivando a primeira equação de Friedmann
em relação ao tempo e utilizando a segunda
equação, obtemos:

$$
\boxed{
\dot{\rho}
+3\frac{\dot a}{a}(\rho+p)=0
}
\tag{3}
$$

Essa equação descreve a evolução da
densidade de energia $\rho(t)$ de um
fluido cosmológico em função da expansão
do Universo e da pressão $p(t)$.
